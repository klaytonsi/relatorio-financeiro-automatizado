## EN

### Problem
Administrative routines often mean merging several spreadsheets by hand, checking for errors and building a summary. Duplicates, empty cells, negative values and mixed date formats are easy to miss.

### Solution
A Python pipeline that runs from start to finish:
1. Generates 4 synthetic spreadsheets with planted errors (CSV and Excel).
2. Consolidates them into one base with Pandas, removing duplicates.
3. Validates each row (negative values, invalid dates, unknown categories) and writes the errors to a separate report.
4. Excel summary by month and category: *in progress*.

### Tools
Python, Pandas, Kaggle (notebook), GitHub. The Excel step will use openpyxl.

### How to run
Open the notebook on Kaggle and run all cells from top to bottom. The data is generated automatically, so no external files are needed.

### Results
- 4 source files consolidated into 160 rows.
- 12 duplicates removed.
- 46 validation errors: 20 missing values, 10 negative amounts, 8 invalid dates, 8 unknown categories.
- Repeated runs gave the same numbers.

### Learnings
Validating data before building summaries prevents wrong totals. Generating data with known errors makes it possible to test whether the validation really catches them.

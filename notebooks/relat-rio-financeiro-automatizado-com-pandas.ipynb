## EN

### Problem
Administrative routines often mean merging several spreadsheets by hand, checking for errors and building a monthly summary. Duplicates, empty cells, negative values and mixed date formats are easy to miss and lead to wrong totals.

### Solution
A Python pipeline that runs from start to finish with no manual files:
1. Generates 4 synthetic spreadsheets (3 CSV and 1 Excel) with planted errors.
2. Consolidates them into one base with Pandas and separates duplicates.
3. Validates each row (missing values, negative amounts, invalid dates, unknown categories) and writes the errors to a separate report.
4. Builds a formatted Excel report with openpyxl (by month, by category and errors) and clean CSVs for a dashboard.

### Tools
Python, Pandas, openpyxl, Kaggle (notebook), GitHub. A Looker Studio dashboard is in progress.

### How to run
Open the notebook on Kaggle and run all cells from top to bottom. The data is generated automatically, so no external files are needed.

### Results
- 4 source files consolidated into 160 rows.
- 12 duplicates separated.
- 58 inconsistencies found: 20 missing values, 12 duplicates, 10 negative amounts, 8 invalid dates, 8 unknown categories.
- Excel report with 3 sheets (by month, by category, errors) and 3 clean CSV summaries.
- Only valid rows enter the summaries, so totals are not distorted by errors.
- Repeated runs gave the same numbers.

### Learnings
Validating data before building summaries prevents wrong totals. Generating data with known errors makes it possible to test whether the validation really catches them. Sharing settings (such as the category list) in one file avoids inconsistencies between scripts.
